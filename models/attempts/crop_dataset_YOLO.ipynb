```python
import zipfile
import os
import random
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Unzip the file
zip_file_path = '/content/Agricultural pests detection.zip'
extract_dir = '/content/pests'

with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)
```

```python
# Count images
image_counts = {}
total_images = 0

for root, dirs, files in os.walk(extract_dir):
    if root != extract_dir:  # Exclude the root directory itself
        class_name = os.path.basename(root)
        image_files = [f for f in files if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
        image_counts[class_name] = len(image_files)
        total_images += len(image_files)

print(f"Total number of images: {total_images}")
print("\nNumber of images per class:")
for class_name, count in image_counts.items():
    print(f"{class_name}: {count}")
```

```python
# Show 10 images per class
print("\nDisplaying 10 images per class:")

for class_name in image_counts.keys():
    print(f"\nClass: {class_name}")
    class_dir = os.path.join(extract_dir, class_name)
    image_files = [f for f in os.listdir(class_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    random_images = random.sample(image_files, min(10, len(image_files)))

    plt.figure(figsize=(15, 5))
    for i, img_file in enumerate(random_images):
        img_path = os.path.join(class_dir, img_file)
        img = mpimg.imread(img_path)
        plt.subplot(1, 10, i + 1)
        plt.imshow(img)
        plt.title(f"Image {i+1}")
        plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
from ultralytics import YOLO
import cv2
import os
import matplotlib.pyplot as plt

# 1) Point this at your folder of images
input_dir = "/content/Agricultural pests detection/earthworms"
os.makedirs(input_dir, exist_ok=True)

# 2) Load a small (fast) YOLOv8 model
model = YOLO("yolov8n.pt")  # or yolov8s.pt, etc.

# 3) Grab the first 10 images
image_files = [f for f in os.listdir(input_dir)
               if f.lower().endswith((".jpg", ".jpeg", ".png"))][-10:]

```

```python
# 4) Run inference, draw boxes, and display
for img_file in image_files:
    img_path = os.path.join(input_dir, img_file)
    results = model(img_path)[0]  # returns a Results object

    # load image for drawing
    img = cv2.imread(img_path)
    for box in results.boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())
        cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)

    # show it
    plt.figure(figsize=(6, 6))
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title(f"Detections in {img_file}")
    plt.axis("off")
    plt.show()
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
```

```python
# --- PARAMETERS ---
data_dir = "/content/pests"
img_size = (224, 224)
batch_size = 32
```

```python
# --- LOAD DATA ---
train_ds = keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=img_size,
    batch_size=batch_size
)
val_ds = keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=img_size,
    batch_size=batch_size
)

# Get class names before prefetching
class_names = train_ds.class_names

# Normalize pixel values
normalization_layer = layers.Rescaling(1./255)
train_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))
val_ds = val_ds.map(lambda x, y: (normalization_layer(x), y))

# Prefetch for performance
train_ds = train_ds.cache().shuffle(1000).prefetch(buffer_size=tf.data.AUTOTUNE)
val_ds = val_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)

# --- DATA AUGMENTATION ---
data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.3),
    layers.RandomContrast(0.3),
    layers.RandomBrightness(0.3),
    layers.RandomTranslation(0.2, 0.2),
])
```
